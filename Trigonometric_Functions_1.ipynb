```python
import math #수학 함수를 사용하기 위해 라이브러리 추가
```

```python
angle_degree = 30 # 각도를 라디안값으로 바꾸기 angle_degree에 일반각을 넣으면 됨.
angle_radian = angle_degree * math.pi / 180 # 각도*π/180
print("angle_radian: ", angle_radian) # 라디안값을 출력
```

```python
math.cos(angle_radian) #라디안값의 코사인 값
```

```python
math.sin(angle_radian) #라디안값의 사인 값
```

```python
math.tan(angle_radian) #라디안값의 탄젠트 값
```

직교좌표를 극좌표로 변환

```python
x,y = 1.0,2.0 #직교좌표의 x값과 y값을 순서대로 넣기
```

```python
r = math.sqrt(x**2 + y**2) # sqrt는 루트, x**2는 x의 2제곱 x**2 + y**2 = r**2의 양변에 루트를 씌운 것
theta_radian = math.atan2(y, x) #atan은 tangent의 역함수 tan(theta) = y/x에서 arctan을 양변에 곱해줌

print(r,theta_radian)
```
