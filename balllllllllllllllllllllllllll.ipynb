```python
import math

# 사용자 입력 받기
y0 = float(input("초기 높이를 입력하세요 (m): "))
v0 = float(input("초기 속도를 입력하세요 (m/s): "))
theta = float(input("초기 각도를 입력하세요 (도): "))

# 상수 정의
g = 9.8
a = -g
pi = math.pi

# 각도를 라디안으로 변환
theta_rad = math.radians(theta)

# 초기 속도의 x, y 성분 계산
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)

# 지면에 닿을 때까지의 시간 계산 (2차 방정식의 해)
A = 0.5 * a
B = v0y
C = y0

discriminant = B**2 - 4 * A * C

if discriminant < 0:
    print("실제 해가 존재하지 않습니다.")
else:
    # C 코드와 동일하게 작은 해 선택
    t_ground = (-B - math.sqrt(discriminant)) / (2 * A)

    # 수평거리
    x = t_ground * v0x

    # 지면 충돌 직전 속도 성분
    vx = v0x
    vy = v0y + a * t_ground

    # 속도 크기
    v = math.sqrt(vx**2 + vy**2)

    # 속도 벡터 각도
    ang = math.degrees(math.atan2(vy, vx))

    # 결과 출력
    print(f"\n지면에 닿을 때 까지의 시간은 : {t_ground:.2f}초")
    print(f"절벽 아래로부터 수평거리는 : {x:.2f}m")
    print(f"포사체가 지면에 부딪치기 직전의 속도 성분은 : {vx:.2f}m/s (수평), {vy:.2f}m/s (수직)")
    print(f"포사체의 최종 속도 크기 : {v:.2f}m/s")
    print(f"속도 벡터가 수평면과 이루는 각도 : {ang:.2f}°")
```
